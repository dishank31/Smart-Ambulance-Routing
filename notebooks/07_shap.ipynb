Model interpretability using SHAP for your trained models.

## Cell 1: Setup

```python
from google.colab import drive
drive.mount('/content/drive')

import os
PROJECT_DIR = '/content/drive/MyDrive/smart-ambulance-ml'

!pip install -q shap matplotlib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib
import shap
import warnings
warnings.filterwarnings('ignore')

print("✓ SHAP ready")
```

## Cell 2: SHAP for Severity Model

```python
print("="*60)
print("SHAP ANALYSIS - SEVERITY")
print("="*60)

# Load data
data = np.load(f'{PROJECT_DIR}/datasets/processed/severity_train.npz', allow_pickle=True)
X_test = data['X_test'][:500]
feature_names = data['feature_names'].tolist()[:20]  # Top 20 features
X_test = X_test[:, :20]

# Load model
model = joblib.load(f'{PROJECT_DIR}/models/severity/xgboost.joblib')

# SHAP explainer
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_test)

# Summary plot
plt.figure(figsize=(10, 8))
shap.summary_plot(shap_values, X_test, feature_names=feature_names, show=False)
plt.title('SHAP - Severity Model (Top 20 Features)')
plt.tight_layout()
plt.savefig(f'{PROJECT_DIR}/results/shap_severity.png', dpi=150, bbox_inches='tight')
plt.show()
print(f"\n✓ Saved to {PROJECT_DIR}/results/shap_severity.png")
```
